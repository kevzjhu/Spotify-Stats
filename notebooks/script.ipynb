```python
import spotipy
from spotipy.oauth2 import SpotifyOAuth
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
spotipy_client_id = os.getenv("SPOTIPY_CLIENT_ID")
spotipy_client_secret = os.getenv("SPOTIPY_CLIENT_SECRET")
spotipy_redirect_uri = os.getenv("SPOTIPY_REDIRECT_URI")
scope = "user-library-read playlist-read-private playlist-read-collaborative"

sp = spotipy.Spotify(
    auth_manager = SpotifyOAuth(
        client_id=spotipy_client_id, 
        client_secret=spotipy_client_secret, 
        redirect_uri=spotipy_redirect_uri, 
        scope=scope,
        open_browser = True
    )
)

user = sp.current_user()
print(f"Logged in as: {user['display_name']} ({user['id']})")
```

```python
# Saved (liked) tracks
results = sp.current_user_saved_tracks()

for idx, item in enumerate(results["items"]):
    track = item["track"]
    print(f"{idx + 1}. {track['name']} by {track['artists'][0]['name']}")
```

```python
# playlists
from IPython.display import JSON
import json

playlists = sp.current_user_playlists(limit = 1)

while playlists:
    for i, playlist in enumerate(playlists['items']):
        print(f"{i + 1 + playlists['offset']:4d} {playlist['uri']} {playlist['name']}")
    if playlists['next']:
        playlists = sp.next(playlists)
    else:
        playlists = None

```

```python
# Replace with your playlist ID or URI
PLAYLIST_ID = "3FvLROJEj2jNSEeJV7i8mh"

# 1. Initial request
results = sp.playlist_items(PLAYLIST_ID)
tracks = results["items"]

# 2. Paginate if playlist has more than 100 items
while results["next"]:
    results = sp.next(results)
    tracks.extend(results["items"])


# 3. Extract track details safely
for i, entry in enumerate(tracks):
    # Retrieve the inner item dictionary
    track = entry.get("item")

    # Guard against nulls (local files or deleted tracks)
    if not track:
        continue

    track_name = track.get("name")
    artist_name = track["artists"][0]["name"] if track.get("artists") else "Unknown"
    track_id = track.get("id")

    print(f"{i + 1:3d}. {track_name} by {artist_name} (ID: {track_id})")
```
